```python
# your code goes here
from scipy.optimize import linprog
```

### Step 3.2: Execute the programming model. (4%)

```python
# your code goes here
obj = [-1, -1, -1]
#      ─┬  ─┬ ┬-| Coefficient for z
#       │   └┤ Coefficient for y
#       └────┤ Coefficient for x

lhs_ineq = [
    [1, 0, 0],  # x1 <= 4000
    [0, 1, 0],  # x2 <= 7000
    [0, 0, 1],  # x3 <= 2500
    [-1, -1, -1]]  # x1 + x2 + x3 >= 12000 (we negate this for <= form)

rhs_ineq = [4000, 7000, 2500,-12000]  # Yellow constraint right side
```

```python
bnd = [(3500, 4000),  # Bounds of x
       (5000, 7000),  # Bounds of y 
       (2100, 2500)]  # Bounds of z
```

```python
opt = linprog(c=obj, A_ub=lhs_ineq, b_ub=rhs_ineq, bounds=bnd,
              method='highs')
```
